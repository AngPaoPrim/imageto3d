```python
# ============================================================
# CELL 5 — HUNYUAN3D-2MV STANDARD + TEXTURE
# DUAL T4
#
# GPU 0 = Shape
# GPU 1 = Texture
# PORT = 8001
# ============================================================

import os
import sys
import time
import subprocess
from pathlib import Path

PYTHON = "/content/hunyuan3d_env/bin/python"

BASE = Path(
    "/content/hunyuan3d_server"
)

SCRIPT = (
    BASE
    / "server.py"
)

LOG = (
    BASE
    / "server.log"
)

PORT = 8001

BASE.mkdir(
    parents=True,
    exist_ok=True
)

print("=" * 70)
print("HUNYUAN3D-2MV STANDARD + TEXTURE")
print("=" * 70)

# ============================================================
# CHECK PYTHON
# ============================================================

if not Path(PYTHON).exists():

    raise FileNotFoundError(
        f"Python not found:\n{PYTHON}"
    )

print(
    "Python:",
    PYTHON
)

subprocess.run(
    [
        PYTHON,
        "--version"
    ],
    check=True
)


# ============================================================
# CHECK SERVER FILE
# ============================================================

if not SCRIPT.exists():

    raise FileNotFoundError(
        f"server.py not found:\n{SCRIPT}"
    )

print(
    "Server:",
    SCRIPT
)

print(
    "Log:",
    LOG
)


# ============================================================
# STOP OLD SERVER
# ============================================================

subprocess.run(
    [
        "pkill",
        "-f",
        "/content/hunyuan3d_server/server.py"
    ],
    stdout=subprocess.DEVNULL,
    stderr=subprocess.DEVNULL,
    check=False
)

time.sleep(2)


# ============================================================
# SERVER CODE
# ============================================================

SERVER_CODE = r'''
import os
import sys
import gc
import uuid
import traceback

from pathlib import Path
from typing import Optional


# ============================================================
# ENVIRONMENT
# ============================================================

os.environ[
    "PYTORCH_CUDA_ALLOC_CONF"
] = "expandable_segments:True"

os.environ[
    "CUDA_MODULE_LOADING"
] = "LAZY"

os.environ[
    "TOKENIZERS_PARALLELISM"
] = "false"


# ============================================================
# CUSTOM RASTERIZER PATH
# ============================================================

RASTERIZER_PATH = (
    "/content/Hunyuan3D-2/"
    "hy3dgen/texgen/custom_rasterizer"
)

REPO_PATH = (
    "/content/Hunyuan3D-2"
)

if RASTERIZER_PATH not in sys.path:

    sys.path.insert(
        0,
        RASTERIZER_PATH
    )

if REPO_PATH not in sys.path:

    sys.path.insert(
        0,
        REPO_PATH
    )


# ============================================================
# IMPORTS
# ============================================================

import torch

from PIL import Image

from fastapi import (
    FastAPI,
    UploadFile,
    File,
    HTTPException,
)

from fastapi.responses import FileResponse

import uvicorn


# ============================================================
# HUNYUAN IMPORTS
# ============================================================

from hy3dgen.shapegen import (
    Hunyuan3DDiTFlowMatchingPipeline
)

from hy3dgen.texgen import (
    Hunyuan3DPaintPipeline
)


# ============================================================
# PATHS
# ============================================================

BASE = Path(
    "/content/hunyuan3d_server"
)

UPLOADS = (
    BASE
    / "uploads"
)

OUTPUTS = (
    BASE
    / "outputs"
)

UPLOADS.mkdir(
    parents=True,
    exist_ok=True
)

OUTPUTS.mkdir(
    parents=True,
    exist_ok=True
)


# ============================================================
# GPU CHECK
# ============================================================

print("=" * 70, flush=True)
print("HUNYUAN3D-2MV STANDARD + TEXTURE", flush=True)
print("=" * 70, flush=True)

print(
    "Python:",
    sys.version,
    flush=True
)

print(
    "Torch:",
    torch.__version__,
    flush=True
)

print(
    "Torch CUDA:",
    torch.version.cuda,
    flush=True
)

print(
    "CUDA available:",
    torch.cuda.is_available(),
    flush=True
)

if not torch.cuda.is_available():

    raise RuntimeError(
        "CUDA is not available."
    )


GPU_COUNT = (
    torch.cuda.device_count()
)

print(
    "GPU COUNT:",
    GPU_COUNT,
    flush=True
)

if GPU_COUNT < 2:

    raise RuntimeError(
        "This server requires 2 GPUs."
    )


for i in range(GPU_COUNT):

    props = (
        torch.cuda
        .get_device_properties(i)
    )

    print(
        f"GPU {i}:",
        props.name,
        "|",
        round(
            props.total_memory
            / 1024**3,
            2
        ),
        "GB",
        flush=True
    )


SHAPE_DEVICE = "cuda:0"
TEXTURE_DEVICE = "cuda:1"


# ============================================================
# LOAD SHAPE MODEL
# ============================================================

print("=" * 70, flush=True)

print(
    "LOADING SHAPE MODEL ON GPU 0...",
    flush=True
)

print("=" * 70, flush=True)


shape_pipeline = (

    Hunyuan3DDiTFlowMatchingPipeline

    .from_pretrained(

        "tencent/Hunyuan3D-2mv",

        subfolder=(
            "hunyuan3d-dit-v2-mv"
        ),

        variant="fp16",

        use_safetensors=True,

        device=SHAPE_DEVICE,
    )
)


print(
    "✅ SHAPE MODEL READY",
    flush=True
)

print(
    "GPU 0 VRAM:",
    round(
        torch.cuda.memory_allocated(0)
        / 1024**3,
        2
    ),
    "GB",
    flush=True
)


# ============================================================
# LOAD TEXTURE MODEL
# ============================================================

print("=" * 70, flush=True)

print(
    "LOADING TEXTURE MODEL...",
    flush=True
)

print("=" * 70, flush=True)


# ------------------------------------------------------------
# Explicitly make sure libc10 / torch shared libraries
# are visible to the process.
# ------------------------------------------------------------

torch_lib = (
    "/content/hunyuan3d_env/"
    "lib/python3.10/site-packages/torch/lib"
)

cuda_lib = (
    "/usr/local/cuda/lib64"
)

ld_paths = [
    torch_lib,
    cuda_lib,
    os.environ.get(
        "LD_LIBRARY_PATH",
        ""
    )
]

os.environ[
    "LD_LIBRARY_PATH"
] = ":".join(
    p
    for p in ld_paths
    if p
)


# ------------------------------------------------------------
# Verify custom rasterizer
# ------------------------------------------------------------

print(
    "Testing custom_rasterizer_kernel...",
    flush=True
)

import custom_rasterizer_kernel

print(
    "custom_rasterizer_kernel: OK",
    flush=True
)

print(
    "Kernel:",
    custom_rasterizer_kernel.__file__,
    flush=True
)


# ============================================================
# LOAD PAINT
# ============================================================

print(
    "Creating Hunyuan3DPaintPipeline...",
    flush=True
)


texture_pipeline = (

    Hunyuan3DPaintPipeline

    .from_pretrained(
        "tencent/Hunyuan3D-2"
    )
)


# ============================================================
# MOVE TEXTURE MODEL TO GPU 1
# ============================================================

print(
    "Moving texture pipeline to GPU 1...",
    flush=True
)

try:

    if hasattr(
        texture_pipeline,
        "to"
    ):

        texture_pipeline.to(
            TEXTURE_DEVICE
        )

        print(
            "Texture pipeline .to(cuda:1) OK",
            flush=True
        )

    else:

        print(
            "WARNING: Texture pipeline has no .to()",
            flush=True
        )

except Exception as e:

    print(
        "WARNING: texture .to(cuda:1) failed:",
        repr(e),
        flush=True
    )


# ------------------------------------------------------------
# Select GPU 1 as current device
# ------------------------------------------------------------

torch.cuda.set_device(
    1
)


print(
    "✅ TEXTURE MODEL READY",
    flush=True
)

print(
    "GPU 1 VRAM:",
    round(
        torch.cuda.memory_allocated(1)
        / 1024**3,
        2
    ),
    "GB",
    flush=True
)


# ============================================================
# RETURN CURRENT DEVICE TO GPU 0
# ============================================================

torch.cuda.set_device(
    0
)


# ============================================================
# API
# ============================================================

app = FastAPI(
    title=(
        "Hunyuan3D-2mv "
        "Dual GPU Texture API"
    )
)


# ============================================================
# HEALTH
# ============================================================

@app.get("/health")
def health():

    return {

        "status":
            "ok",

        "model":
            "tencent/Hunyuan3D-2mv",

        "mode":
            "multi-view",

        "textured":
            True,

        "shape_gpu":
            torch.cuda.get_device_name(0),

        "texture_gpu":
            torch.cuda.get_device_name(1),

        "shape_vram_gb":
            round(
                torch.cuda.memory_allocated(0)
                / 1024**3,
                2
            ),

        "texture_vram_gb":
            round(
                torch.cuda.memory_allocated(1)
                / 1024**3,
                2
            ),
    }


# ============================================================
# GENERATE MULTI-VIEW
# ============================================================

@app.post("/generate_multi")
async def generate_multi(

    front: Optional[
        UploadFile
    ] = File(None),

    left: Optional[
        UploadFile
    ] = File(None),

    back: Optional[
        UploadFile
    ] = File(None),

    right: Optional[
        UploadFile
    ] = File(None),
):

    incoming = {

        "front":
            front,

        "left":
            left,

        "back":
            back,

        "right":
            right,
    }


    images = {}

    temp_files = []


    try:

        # ====================================================
        # LOAD IMAGES
        # ====================================================

        for role, upload in incoming.items():

            if upload is None:
                continue


            filename = Path(
                upload.filename
                or f"{role}.png"
            ).name


            path = (

                UPLOADS
                /

                (
                    uuid.uuid4().hex[:8]
                    +
                    "_"
                    +
                    filename
                )
            )


            path.write_bytes(
                await upload.read()
            )

            temp_files.append(
                path
            )


            images[role] = (

                Image.open(
                    path
                )
                .convert("RGBA")
            )


        # ====================================================
        # VALIDATION
        # ====================================================

        if len(images) < 2:

            raise HTTPException(

                status_code=400,

                detail=(
                    "At least 2 views "
                    "are required."
                )
            )


        if "front" not in images:

            raise HTTPException(

                status_code=400,

                detail=(
                    "front image is required."
                )
            )


        print("=" * 70, flush=True)

        print(
            "MULTI-VIEW SHAPE",
            flush=True
        )

        print(
            "Views:",
            list(images.keys()),
            flush=True
        )


        # ====================================================
        # SHAPE ON GPU 0
        # ====================================================

        print(
            "Using GPU 0 for SHAPE...",
            flush=True
        )

        torch.cuda.set_device(
            0
        )


        with torch.inference_mode():

            mesh = (

                shape_pipeline(

                    image=images,

                    num_inference_steps=50,

                    octree_resolution=380,

                    num_chunks=20000,

                    generator=(
                        torch.Generator(
                            device="cuda:0"
                        )
                        .manual_seed(
                            12345
                        )
                    ),

                    output_type="trimesh",

                )[0]

            )


        print(
            "✅ SHAPE GENERATED",
            flush=True
        )


        print(
            "GPU 0 VRAM:",
            round(
                torch.cuda.memory_allocated(0)
                / 1024**3,
                2
            ),
            "GB",
            flush=True
        )


        # ====================================================
        # TEXTURE ON GPU 1
        # ====================================================

        print("=" * 70, flush=True)

        print(
            "TEXTURE GENERATION",
            flush=True
        )

        print(
            "Using GPU 1...",
            flush=True
        )


        torch.cuda.set_device(
            1
        )


        front_image = (
            images["front"]
        )


        with torch.inference_mode():

            textured_mesh = (

                texture_pipeline(

                    mesh,

                    image=front_image

                )

            )


        print(
            "✅ TEXTURE GENERATED",
            flush=True
        )


        print(
            "GPU 1 VRAM:",
            round(
                torch.cuda.memory_allocated(1)
                / 1024**3,
                2
            ),
            "GB",
            flush=True
        )


        # ====================================================
        # SAVE GLB
        # ====================================================

        filename = (

            "hunyuan_textured_"

            +
            uuid.uuid4().hex[:10]

            +
            ".glb"
        )


        output = (
            OUTPUTS
            / filename
        )


        textured_mesh.export(
            str(output)
        )


        print("=" * 70, flush=True)

        print(
            "✅ FINAL TEXTURED GLB",
            flush=True
        )

        print(
            "FILE:",
            output,
            flush=True
        )


        # ====================================================
        # CLEANUP
        # ====================================================

        del mesh

        del textured_mesh

        for image in images.values():

            try:
                image.close()
            except Exception:
                pass


        del images

        gc.collect()

        torch.cuda.empty_cache()


        torch.cuda.set_device(
            0
        )


        return {

            "status":
                "completed",

            "filename":
                filename,

            "result":
                "/result/"
                +
                filename,

            "textured":
                True,

            "shape_gpu":
                0,

            "texture_gpu":
                1,

        }


    except HTTPException:

        raise


    except Exception as error:

        traceback.print_exc()

        raise HTTPException(

            status_code=500,

            detail=str(error)
        )


    finally:

        for path in temp_files:

            try:
                path.unlink()
            except Exception:
                pass


# ============================================================
# RESULT
# ============================================================

@app.get(
    "/result/{filename}"
)
def result(
    filename: str
):

    safe = Path(
        filename
    ).name


    path = (
        OUTPUTS
        /
        safe
    )


    if not path.exists():

        raise HTTPException(

            status_code=404,

            detail="Result not found"
        )


    return FileResponse(

        str(path),

        media_type=(
            "model/gltf-binary"
        ),

        filename=safe
    )


# ============================================================
# START
# ============================================================

print("=" * 70, flush=True)

print(
    "HUNYUAN3D DUAL-GPU API :8001",
    flush=True
)

print("=" * 70, flush=True)


uvicorn.run(

    app,

    host="127.0.0.1",

    port=8001,

    log_level="info"
)
'''


# ============================================================
# WRITE SERVER
# ============================================================

SCRIPT.write_text(
    SERVER_CODE
)

print(
    "Server script written."
)


# ============================================================
# SYNTAX
# ============================================================

subprocess.run(

    [
        PYTHON,
        "-m",
        "py_compile",
        str(SCRIPT)
    ],

    check=True
)

print(
    "SYNTAX: OK"
)


# ============================================================
# SERVER ENV
# ============================================================

env = os.environ.copy()

env[
    "PYTHONUNBUFFERED"
] = "1"

env[
    "PYTORCH_CUDA_ALLOC_CONF"
] = (
    "expandable_segments:True"
)

env[
    "CUDA_MODULE_LOADING"
] = "LAZY"

env[
    "TOKENIZERS_PARALLELISM"
] = "false"


# IMPORTANT:
# Make torch libraries visible to custom_rasterizer_kernel.

env[
    "LD_LIBRARY_PATH"
] = (

    "/content/hunyuan3d_env/"
    "lib/python3.10/site-packages/torch/lib:"

    "/usr/local/cuda/lib64:"

    + env.get(
        "LD_LIBRARY_PATH",
        ""
    )
)


# Also expose rasterizer path.

env[
    "PYTHONPATH"
] = (

    "/content/Hunyuan3D-2/"
    "hy3dgen/texgen/custom_rasterizer:"

    "/content/Hunyuan3D-2:"

    + env.get(
        "PYTHONPATH",
        ""
    )
)


# ============================================================
# START BACKGROUND SERVER
# ============================================================

log_handle = open(
    LOG,
    "w",
    buffering=1
)


process = subprocess.Popen(

    [
        PYTHON,
        "-u",
        str(SCRIPT)
    ],

    cwd="/content",

    env=env,

    stdout=log_handle,

    stderr=subprocess.STDOUT,

    start_new_session=True
)


log_handle.close()


print()
print("=" * 70)
print("✅ SERVER STARTED IN BACKGROUND")
print("=" * 70)

print(
    "PID:",
    process.pid
)

print(
    "PORT:",
    PORT
)

print(
    "LOG:",
    LOG
)

print()
print(
    "GPU 0 = Shape"
)

print(
    "GPU 1 = Texture"
)

print()
print(
    "ตอนนี้รัน CELL 6 ได้เลย"
)
```

```python
# ============================================================
# CELL 6 — WAIT FOR API
# ============================================================

import time
import requests
import subprocess
from pathlib import Path

PORT = 8001

LOG = Path(
    "/content/hunyuan3d_server/server.log"
)

TIMEOUT = 1800

START = time.time()

print("=" * 70)
print("WAITING FOR HUNYUAN3D-2MV API")
print("=" * 70)

last_size = 0


while True:

    # ========================================================
    # HEALTH CHECK
    # ========================================================

    try:

        response = requests.get(
            f"http://127.0.0.1:{PORT}/health",
            timeout=5
        )

        if response.status_code == 200:

            print()
            print("=" * 70)
            print("✅ HUNYUAN3D-2MV API READY")
            print("=" * 70)

            print(
                response.json()
            )

            print()
            print(
                "Generate:",
                f"http://127.0.0.1:{PORT}/generate_multi"
            )

            break

    except Exception:
        pass


    # ========================================================
    # CHECK SERVER PROCESS
    # ========================================================

    proc = subprocess.run(

        [
            "bash",
            "-lc",

            "pgrep -af "
            "'/content/hunyuan3d_server/server.py' "
            "|| true"
        ],

        capture_output=True,

        text=True,

        check=False
    )


    if not proc.stdout.strip():

        print()
        print("=" * 70)
        print("❌ HUNYUAN SERVER STOPPED")
        print("=" * 70)

        if LOG.exists():

            print(
                LOG.read_text(
                    errors="replace"
                )[-40000:]
            )

        raise RuntimeError(
            "Hunyuan3D server stopped."
        )


    # ========================================================
    # SHOW NEW LOG
    # ========================================================

    if LOG.exists():

        try:

            size = (
                LOG.stat()
                .st_size
            )

            if size != last_size:

                text = LOG.read_text(
                    errors="replace"
                )

                print(
                    text[-4000:]
                )

                last_size = size

        except Exception:
            pass


    # ========================================================
    # TIMEOUT
    # ========================================================

    elapsed = (
        time.time()
        - START
    )


    if elapsed >= TIMEOUT:

        print()
        print("=" * 70)
        print("❌ SERVER STARTUP TIMEOUT")
        print("=" * 70)

        if LOG.exists():

            print(
                LOG.read_text(
                    errors="replace"
                )[-40000:]
            )

        raise TimeoutError(
            f"Server did not become ready "
            f"within {TIMEOUT} seconds."
        )


    time.sleep(5)
```

```python
# ============================================================
# PATCH — multiview_utils.py
# ============================================================

from pathlib import Path

FILE = Path(
    "/content/Hunyuan3D-2/hy3dgen/texgen/utils/multiview_utils.py"
)

print("=" * 70)
print("PATCHING multiview_utils.py")
print("=" * 70)

text = FILE.read_text(
    errors="replace"
)

old = (
    "pipeline = DiffusionPipeline.from_pretrained(\n"
    "            multiview_ckpt_path,\n"
    "            custom_pipeline=custom_pipeline_path, torch_dtype=torch.float16)"
)

new = (
    "pipeline = DiffusionPipeline.from_pretrained(\n"
    "            multiview_ckpt_path,\n"
    "            custom_pipeline=custom_pipeline_path,\n"
    "            torch_dtype=torch.float16,\n"
    "            trust_remote_code=True)"
)

if "trust_remote_code=True" in text:

    print(
        "✅ Already patched"
    )

elif old in text:

    FILE.write_text(
        text.replace(
            old,
            new
        )
    )

    print(
        "✅ PATCH APPLIED"
    )

else:

    # ใช้ line-based fallback
    lines = text.splitlines()

    patched = False

    for i, line in enumerate(lines):

        if (
            "custom_pipeline=custom_pipeline_path"
            in line
            and
            "torch_dtype=torch.float16"
            in line
        ):

            lines[i] = (
                "            custom_pipeline=custom_pipeline_path,"
            )

            lines.insert(
                i + 1,
                "            torch_dtype=torch.float16,"
            )

            lines.insert(
                i + 2,
                "            trust_remote_code=True)"
            )

            patched = True
            break

    if not patched:

        raise RuntimeError(
            "ไม่สามารถหา block สำหรับ patch ได้"
        )

    FILE.write_text(
        "\n".join(lines)
        + "\n"
    )

    print(
        "✅ PATCH APPLIED (fallback)"
    )


# ============================================================
# SHOW RESULT
# ============================================================

print()
print("=" * 70)
print("VERIFY")
print("=" * 70)

lines = FILE.read_text(
    errors="replace"
).splitlines()

for i, line in enumerate(lines):

    if "DiffusionPipeline.from_pretrained" in line:

        for n in range(
            i,
            min(
                len(lines),
                i + 8
            )
        ):

            print(
                f"{n + 1}: {lines[n]}"
            )

        break

if "trust_remote_code=True" not in "\n".join(lines):

    raise RuntimeError(
        "Patch verification failed"
    )

print()
print(
    "✅ trust_remote_code=True VERIFIED"
)
```
