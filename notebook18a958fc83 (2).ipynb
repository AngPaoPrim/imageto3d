```python
# ============================================================
# CELL 5 — START HUNYUAN3D-2MV + TEXTURE SERVER
# BACKGROUND MODE
# PORT 8001
# ============================================================

import os
import time
import subprocess
from pathlib import Path

PYTHON = "/content/hunyuan3d_env/bin/python"

SERVER_DIR = Path(
    "/content/hunyuan3d_server"
)

SCRIPT = (
    SERVER_DIR
    / "server.py"
)

LOG = (
    SERVER_DIR
    / "server.log"
)

PORT = 8001


# ============================================================
# CHECK FILES
# ============================================================

print("=" * 70)
print("HUNYUAN3D-2MV SERVER START")
print("=" * 70)

print(
    "Python:",
    PYTHON
)

print(
    "Server:",
    SCRIPT
)

print(
    "Log:",
    LOG
)

if not Path(PYTHON).exists():

    raise FileNotFoundError(
        f"Python environment not found:\n{PYTHON}"
    )

if not SCRIPT.exists():

    raise FileNotFoundError(
        f"Server file not found:\n{SCRIPT}"
    )


# ============================================================
# CHECK PYTHON
# ============================================================

subprocess.run(
    [
        PYTHON,
        "--version"
    ],
    check=True
)


# ============================================================
# SYNTAX CHECK
# ============================================================

print()
print(
    "Checking server syntax..."
)

syntax = subprocess.run(
    [
        PYTHON,
        "-m",
        "py_compile",
        str(SCRIPT)
    ],
    capture_output=True,
    text=True
)

if syntax.returncode != 0:

    print(
        syntax.stdout
    )

    print(
        syntax.stderr
    )

    raise RuntimeError(
        "server.py syntax error"
    )

print(
    "SYNTAX: OK"
)


# ============================================================
# STOP OLD INSTANCE
# ============================================================

print()
print(
    "Stopping old Hunyuan server..."
)

subprocess.run(
    [
        "pkill",
        "-f",
        "/content/hunyuan3d_server/server.py"
    ],
    stdout=subprocess.DEVNULL,
    stderr=subprocess.DEVNULL,
    check=False
)

time.sleep(3)


# ============================================================
# ENVIRONMENT
# ============================================================

env = os.environ.copy()

env[
    "PYTHONUNBUFFERED"
] = "1"

env[
    "PYTORCH_CUDA_ALLOC_CONF"
] = (
    "expandable_segments:True"
)

env[
    "CUDA_MODULE_LOADING"
] = "LAZY"

env[
    "TOKENIZERS_PARALLELISM"
] = "false"


# ============================================================
# LOG FILE
# ============================================================

log_handle = open(
    LOG,
    "w",
    buffering=1
)


# ============================================================
# START BACKGROUND PROCESS
# ============================================================

process = subprocess.Popen(

    [
        PYTHON,
        "-u",
        str(SCRIPT)
    ],

    cwd="/content",

    env=env,

    stdout=log_handle,

    stderr=subprocess.STDOUT,

    start_new_session=True
)


# ============================================================
# CLOSE PARENT COPY OF LOG
# ============================================================

log_handle.close()


# ============================================================
# DONE
# ============================================================

print()
print("=" * 70)
print("✅ HUNYUAN3D SERVER STARTED")
print("=" * 70)

print(
    "PID:",
    process.pid
)

print(
    "PORT:",
    PORT
)

print(
    "LOG:",
    LOG
)

print()
print(
    "Server กำลังโหลดโมเดลเบื้องหลัง..."
)

print(
    "Cell นี้จบแล้ว ค่อยรัน CELL 6"
)
```

```python
# ============================================================
# CELL 6 — WAIT FOR HUNYUAN3D-2MV API
# ============================================================

import time
import requests
import subprocess
from pathlib import Path

PORT = 8001

LOG = Path(
    "/content/hunyuan3d_server/server.log"
)

START = time.time()

TIMEOUT = 1800


def is_server_running():

    result = subprocess.run(
        [
            "bash",
            "-lc",
            "pgrep -af '/content/hunyuan3d_server/server.py' || true"
        ],
        capture_output=True,
        text=True,
        check=False
    )

    return bool(
        result.stdout.strip()
    )


print("=" * 70)
print("WAITING FOR HUNYUAN3D-2MV")
print("=" * 70)

last_line = ""


while True:

    # ========================================================
    # HEALTH CHECK
    # ========================================================

    try:

        response = requests.get(
            f"http://127.0.0.1:{PORT}/health",
            timeout=5
        )

        if response.status_code == 200:

            print()
            print("=" * 70)
            print("✅ HUNYUAN3D-2MV API READY")
            print("=" * 70)

            print(
                response.json()
            )

            print()
            print(
                "Endpoint:",
                f"http://127.0.0.1:{PORT}/generate_multi"
            )

            break

    except Exception:
        pass


    # ========================================================
    # PROCESS CHECK
    # ========================================================

    if not is_server_running():

        print()
        print("=" * 70)
        print("❌ SERVER PROCESS NOT RUNNING")
        print("=" * 70)

        if LOG.exists():

            print()
            print(
                LOG.read_text(
                    errors="replace"
                )[-30000:]
            )

        raise RuntimeError(
            "Hunyuan3D server stopped before API became ready."
        )


    # ========================================================
    # SHOW LATEST LOG LINE
    # ========================================================

    if LOG.exists():

        try:

            lines = LOG.read_text(
                errors="replace"
            ).splitlines()

            if lines:

                current = lines[-1]

                if current != last_line:

                    print(
                        "[HUNYUAN]",
                        current
                    )

                    last_line = current

        except Exception:
            pass


    # ========================================================
    # TIMEOUT
    # ========================================================

    elapsed = (
        time.time()
        - START
    )

    if elapsed > TIMEOUT:

        print()
        print("=" * 70)
        print("❌ STARTUP TIMEOUT")
        print("=" * 70)

        if LOG.exists():

            print(
                LOG.read_text(
                    errors="replace"
                )[-30000:]
            )

        raise TimeoutError(
            "Hunyuan3D startup exceeded 1800 seconds."
        )


    time.sleep(5)
```
